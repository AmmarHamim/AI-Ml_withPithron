## ১. Supervised Learning কী?

Supervised Learning হলো সেই ML পদ্ধতি যেখানে মডেলকে labeled data দিয়ে train করা হয় — অর্থাৎ প্রতিটি input-এর সাথে সঠিক output দেওয়া থাকে। মডেল সেই উদাহরণ থেকে শিখে নতুন data-তে prediction করে। যেমন: ১০০০টি ছবি দিলে যেখানে প্রতিটিতে লেখা আছে "cat" বা "dog" — মডেল শিখে নতুন ছবি দেখলে বলতে পারে কোনটা কী।

## 2. Regression vs Classification

### বিষয়	   Regression	     Classification

Example	House price, temperature	Spam/Not spam, Cat/Dog

## ৩. House Price Prediction — কোন ধরনের problem?

এটি Regression problem। কারণ output হলো একটি continuous সংখ্যা (যেমন ৳৪৫,৩০,০০০) — কোনো নির্দিষ্ট category নয়। বাড়ির size, location, rooms সংখ্যা দিয়ে একটি নির্দিষ্ট মূল্য predict করতে হয়, যা যেকোনো সংখ্যা হতে পারে।


## ৪. Spam Email Detection — কোন ধরনের ML problem?

এটি Supervised Classification problem। কারণ:

Output দুটি নির্দিষ্ট class: Spam বা Not Spam
Training data-তে প্রতিটি email আগে থেকেই labeled থাকে

## ৫. Unsupervised Learning এ labels কেন থাকে না?

কারণ Unsupervised Learning-এর উদ্দেশ্যই হলো ডেটার ভেতর থেকে নিজে নিজে pattern বা structure খুঁজে বের করা — আগে থেকে কোনো সঠিক উত্তর জানা নেই। Labels দেওয়া সম্ভব না কারণ:

ডেটা অনেক বেশি বা labeling ব্যয়বহুল
আমরা জানি না কতগুলো বা কী ধরনের group আছে
লক্ষ্যই হলো অজানা pattern আবিষ্কার করা

উদাহরণ: একটি দোকানের ১০ লাখ customer-এর purchase history দিলে, মডেল নিজে নিজে similar customer-দের group করে ফেলবে — আমরা আগে থেকে group নির্ধারণ করে দিইনি।

## ১. Feature Engineering কী?

Raw data থেকে মডেলের জন্য উপযুক্ত input variables (features) তৈরি বা বাছাই করার প্রক্রিয়া। মডেল সরাসরি বুঝতে পারে না বলে আমাদের ডেটাকে অর্থবহ সংখ্যায় রূপান্তর করতে হয়।
উদাহরণ: "জন্মতারিখ" থেকে সরাসরি কাজ না করে "বয়স" বের করে দেওয়া — এটি feature engineering। একইভাবে "সকাল/দুপুর/রাত" বের করা timestamp থেকে।

## ২. Train-Test Split কেন প্রয়োজন?

মডেল যে ডেটায় train হয়, সেই একই ডেটায় test করলে সে "মুখস্থ" উত্তর দেবে — আসল performance বোঝা যাবে না। তাই ডেটাকে দুভাগ করা হয়:

Training set (৭০–৮০%): মডেল এই ডেটা থেকে শেখে
Test set (২০–৩০%): মডেল এই ডেটা আগে কখনো দেখেনি, তাই এখানে performance আসল দুনিয়ার মতো

এটি না করলে Overfitting ধরা পড়বে না।

## ৩. Data Preprocessing এর ৩টি কাজ

i. Missing Value Handle করা — অনেক সময় dataset-এ কিছু ঘর ফাঁকা থাকে। সেগুলো mean/median দিয়ে fill করা হয় অথবা সেই row বাদ দেওয়া হয়।

ii. Normalization / Scaling — বিভিন্ন feature-এর range আলাদা হলে (যেমন বয়স: ১–১০০, বেতন: ১০,০০০–১,০০,০০০) মডেল বিভ্রান্ত হয়। সব feature-কে একই scale-এ (যেমন ০ থেকে ১) নিয়ে আসা হয়।

iii. Categorical Encoding — মডেল text বোঝে না। তাই "পুরুষ/মহিলা" বা "ঢাকা/চট্টগ্রাম" এর মতো text data-কে সংখ্যায় রূপান্তর করতে হয় — যেমন One-Hot Encoding।

## ৪. Hyperparameter এবং Parameter এর মধ্যে পার্থক্য

Parameter হলো সেই মান যা মডেল নিজে training-এর সময় ডেটা থেকে শেখে। যেমন Neural Network-এর weights ও biases — এগুলো training চলাকালীন ধীরে ধীরে update হয়, মানুষকে এগুলো সেট করতে হয় না।

Hyperparameter হলো সেই সেটিং যা training শুরুর আগে মানুষ নিজে নির্ধারণ করে দেয়। যেমন learning rate, number of layers, বা epochs কতটা হবে — এগুলো মডেল নিজে শেখে না, বরং মডেলের শেখার প্রক্রিয়াকে নিয়ন্ত্রণ করে।
সহজ কথায়, Parameter হলো মডেল যা শেখে এবং Hyperparameter হলো শেখার আগে আমরা মডেলকে যে নির্দেশনা দিই।

## ৫. Model Drift বলতে কী বুঝায়?

সময়ের সাথে real-world ডেটার pattern পরিবর্তন হওয়ার কারণে মডেলের performance ধীরে ধীরে খারাপ হয়ে যাওয়াকে Model Drift বলে।
দুই ধরনের drift আছে:

Data Drift: Input data-র distribution পরিবর্তন হয়। যেমন করোনার পর মানুষের কেনাকাটার অভ্যাস বদলে গেছে — আগের ডেটায় trained মডেল আর কাজ করে না।
Concept Drift: Input ও Output-এর সম্পর্কই পরিবর্তন হয়। যেমন ৫ বছর আগের spam email-এর ধরন আর এখনকার ধরন আলাদা।

এজন্য production-এ model-কে নিয়মিত monitor ও retrain করতে হয়।

## ১. Simple Linear Regression কী predict করে?

Simple Linear Regression দুটি variable-এর মধ্যে সরলরৈখিক সম্পর্ক ব্যবহার করে একটি continuous সংখ্যা predict করে। একটি input variable (x) দিয়ে একটি output variable (y) predict করা হয়।উদাহরণ: পড়াশোনার ঘণ্টা (x) দিয়ে পরীক্ষার নম্বর (y) predict করা।

## ২. Independent Variable এবং Dependent Variable এর পার্থক্য

Independent Variable হলো সেই variable যা আমরা input হিসেবে দিই — এটি অন্য কিছুর উপর নির্ভর করে না। একে x দিয়ে প্রকাশ করা হয়। যেমন বাড়ির আয়তন (square feet)।
Dependent Variable হলো সেই variable যা independent variable-এর উপর নির্ভর করে পরিবর্তন হয় — এটিই আমরা predict করতে চাই। একে y দিয়ে প্রকাশ করা হয়। যেমন বাড়ির দাম।
সহজ কথায়, x পরিবর্তন হলে y পরিবর্তন হয় — তাই y নির্ভরশীল।

## ৩. y = mx + b equation এ m এবং b কী বোঝায়?

m হলো slope বা gradient — এটি বোঝায় x একটু বাড়লে y কতটুকু বাড়বে বা কমবে। যেমন m = 2 মানে x এক ঘণ্টা বাড়লে y দুই নম্বর বাড়বে।
b হলো intercept — এটি বোঝায় x = 0 হলে y-এর মান কত হবে, অর্থাৎ line টি y-axis কে কোথায় ছেদ করে। যেমন b = 10 মানে একদম না পড়লেও ১০ নম্বর পাওয়া যাবে।

## ৪. Best Fit Line বলতে কী বুঝায়?

Best Fit Line হলো সেই সরলরেখা যা ডেটার সব point-এর যতটা সম্ভব কাছ দিয়ে যায় — অর্থাৎ প্রতিটি actual value এবং predicted value-এর মধ্যে পার্থক্য (error) সবমিলিয়ে সর্বনিম্ন হয়।
এই error-কে Residual বলে এবং সব residual-এর square-এর যোগফল minimize করার পদ্ধতিকে বলে Ordinary Least Squares (OLS)। এই পদ্ধতিতেই best fit line খুঁজে বের করা হয়।

## ১. Cost Function কেন দরকার?
Model কতটা ভালো বা খারাপ prediction করছে সেটা একটি সংখ্যায় পরিমাপ করার জন্য Cost Function দরকার। Training-এর সময় model বারবার prediction করে এবং Cost Function সেই prediction কতটা ভুল তা হিসাব করে একটি সংখ্যা দেয়। সেই সংখ্যা দেখে model নিজেকে উন্নত করে।
Cost Function ছাড়া model জানতে পারবে না সে ভালো করছে না খারাপ করছে — তাই শেখার কোনো দিকনির্দেশনা থাকবে না।

## ২. Prediction Error কী?
Prediction Error হলো model-এর predicted value এবং actual value-এর মধ্যে পার্থক্য।
Error = Actual Value − Predicted Value
যেমন বাড়ির আসল দাম ৫০ লাখ কিন্তু model predict করেছে ৪৩ লাখ — তাহলে error হলো ৭ লাখ। এই error যত বড়, model তত বেশি ভুল করছে।

## ৩. Model error কমানো কেন গুরুত্বপূর্ণ?
Error বেশি থাকলে model real-world-এ কাজের হবে না — ভুল prediction দেবে যা ক্ষতিকর হতে পারে। যেমন রোগ নির্ণয়ে ভুল prediction মানুষের জীবন বিপদে ফেলতে পারে, বা loan approval-এ ভুল সিদ্ধান্ত আর্থিক ক্ষতি করতে পারে।
Error কমানো মানে model-কে actual data-র যত কাছাকাছি নিয়ে যাওয়া — ফলে নতুন অজানা data-তেও সঠিক prediction আসে।

## ৪. Large error হলে Cost Function এর value কেমন হবে?

Large error হলে Cost Function-এর value বেশি হবে। কারণ Cost Function মূলত সব error-এর উপর নির্ভর করে হিসাব করা হয়। সবচেয়ে প্রচলিত Cost Function হলো Mean Squared Error (MSE) যেখানে প্রতিটি error-কে square করে গড় নেওয়া হয়।
MSE = (1/n) × Σ(Actual − Predicted)²
Error বড় হলে square করার পর আরো বড় হয়, ফলে Cost-এর value অনেক বেশি হয়। Cost-এর value বেশি মানে model এখনো অনেক দুর্বল।

## ৫. Cost কমানোর process এর নাম কী?
Cost কমানোর process-এর নাম Gradient Descent। এটি একটি optimization algorithm যা বারবার model-এর parameter (weight ও bias) একটু একটু করে adjust করে Cost Function-এর value কমিয়ে আনে।
পাহাড়ের চূড়া থেকে সবচেয়ে খাড়া ঢাল বেয়ে নিচে নামার মতো — Gradient Descent প্রতিটি step-এ সেই দিকে যায় যেদিকে Cost সবচেয়ে দ্রুত কমে। এই process বারবার চলতে থাকে যতক্ষণ না Cost সর্বনিম্নে পৌঁছায়।

```python
# sklearn ব্যবহার করে একটি simple linear regression model তৈরি করো


from sklearn.linear_model import LinearRegression
import numpy as np

# Sample data
X = np.array([[1000], [1500], [2000], [2500], [3000]])  # house size (sq ft)
y = np.array([200000, 300000, 400000, 500000, 600000])  # house price (৳)

# Model তৈরি ও train করা
model = LinearRegression()
model.fit(X, y)

# Prediction
new_house = np.array([[1800]])
predicted_price = model.predict(new_house)

print(f"Slope (m): {model.coef_[0]}")
print(f"Intercept (b): {model.intercept_}")
print(f"1800 sq ft বাড়ির predicted price: ৳{predicted_price[0]:,.0f}")
```
