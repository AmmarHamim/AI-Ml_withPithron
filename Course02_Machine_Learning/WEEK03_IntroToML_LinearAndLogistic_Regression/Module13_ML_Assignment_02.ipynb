Implement **Simple Linear Regression** using **only NumPy** (no Scikit-Learn allowed):
- Compute slope (`m`) and intercept (`c`) using the Batch Gradient Descent
- Predict values for the test set
- Print the learned `m` and `c` values

Use `Square_Footage` as feature (X) and `House_Price` as target (y).

# Question 3

```python
X_new_train=X_train['Square_Footage'].values
X_new_test=X_test['Square_Footage'].values
y_new_train=y_train.values
y_new_test=y_test.values
```

```python
def compute_cost(X,y,w,b):
  m=X.shape[0]
  cost=0.0

  pred=w*X+b
  error=pred-y
  error_sqr=error**2
  cost=np.sum(error_sqr)


  cost = cost / (2 * m)

  return cost
```

```python
print(compute_cost(X_new_train,y_new_train,w=0,b=0))
print(compute_cost(X_new_train,y_new_train,w=17,b=0))
```

```python
def calculate_gradient(X,y,w,b):
  m=X.shape[0]
  dj_dw=0.0
  dj_db=0.0

  pred=w*X+b
  error=pred-y
  dj_dw=np.sum(error*X)
  dj_db=np.sum(error)

  return dj_dw/m, dj_db/m
```

```python
print(calculate_gradient(X_new_train,y_new_train,w=0,b=0))
print(calculate_gradient(X_new_train,y_new_train,w=19,b=0))
```

```python
def gradient_descent(X,y,w_input,b_input,max_iter,alpha):


  w = w_input
  b = b_input
  cost_memo=[]

  for i in range(max_iter):

    dj_dw,dj_db=calculate_gradient(X,y,w,b)
    w = w - alpha * dj_dw
    b = b - alpha * dj_db

    cost=compute_cost(X,y,w,b)
    cost_memo.append(cost)

    if i % 100 == 0:
      print(f"Iteration {i}: Cost {cost:0.4f}, w: {w:0.4f}, b: {b:0.4f}")


  return w,b
```

```python
w_init=0
b_init=0
MAX_ITER=1000
ALPHA=0.00000001
w_final,b_final=gradient_descent(X_new_train,y_new_train,w_init,b_init,MAX_ITER,ALPHA)
```

```python
print(w_final,b_final)
```

```python
def predict(X,w,b):
  return w*X+b
```

```python
y_pred=predict(X_new_test,w_final,b_final)
y_real=y_new_test

for i in range(10):
  print(f"Predicted = {y_pred[i]:0.4f}  ---->  Actual value = {y_real[i]:0.4f}")
```

Build a **ColumnTransformer** that applies:

```python
numeric_features=['Square_Footage','Num_Bedrooms','Num_Bathrooms']
categoric_features=['Neighborhood_Quality']

preprocessor=ColumnTransformer(
    transformers=[
        ('std_scaler',StandardScaler(),numeric_features),
        ('ohe',OneHotEncoder(handle_unknown='ignore',sparse_output=False),categoric_features)
    ],
    remainder='drop',
    verbose_feature_names_out=False
)

preprocessor.set_output(transform='pandas')
```

```python
X_train_transformed=preprocessor.fit_transform(X_train)
X_test_transformed=preprocessor.transform(X_test)

X_train_transformed
```

## Question 5 (20 Marks)

Build a complete **Pipeline** using Scikit-Learn that includes:
- The `ColumnTransformer`
- `SGDRegressor` as the final estimator
- Train the pipeline and evaluate using RMSE and R² score
- Print predicted vs actual values for the first 10 test samples

# Question 5

```python
ETA=0.01
MAX_ITER=3000
ALPHA=0.001


sgd_pipeline=Pipeline(
    steps=[
        ('preprocessor',preprocessor),
        ('model',SGDRegressor(
            eta0=ETA,
            max_iter=MAX_ITER,
            alpha=ALPHA,
            loss='squared_error',
            learning_rate='constant',
            random_state=42
        ))
    ]
)

sgd_pipeline.fit(X_train,y_train)
```

```python
y_pred_sgd=sgd_pipeline.predict(X_test)
y_real=y_test.values

rmse=root_mean_squared_error(y_real,y_pred_sgd)
r2=r2_score(y_real,y_pred_sgd)

print("SGDRegressor Pipeline Results")
print(f"  RMSE : {rmse:.4f}")
print(f"  R²   : {r2:.4f}")
```

```python
print("\nFirst 10 Predictions vs Actual:")

for i in range(10):
  print(f"Sample : {i}------> Predicted = {y_pred_sgd[i]}------> Actual = {y_real[i]}")
```

Implement **Multiple Linear Regression** using **only NumPy**:
- Pick random 100 datas from the dataset
- Use the Normal Equation: `θ = (XᵀX)⁻¹ Xᵀy`
- Use `Square_Footage`, `Num_Bedrooms`, and `Num_Bathrooms` as features
- Print the learned coefficients (θ values)

#

# Question 7

documentation used: https://medium.com/analytics-vidhya/multiple-linear-regression-from-scratch-using-only-numpy-98fc010a1926

```python
sample_df=df.sample(100,random_state=42)
X=sample_df[['Square_Footage','Num_Bedrooms','Num_Bathrooms']].values
y=sample_df['House_Price'].values

X=np.c_[np.ones(X.shape[0]),X] #adding bias

def normal_equation(X, Y):
    beta = np.dot((np.linalg.inv(np.dot(X.T,X))), np.dot(X.T,Y))

    return beta

def predict(X_test, beta):
    return np.dot(X_test, beta)
```

```python
beta = normal_equation(X,y)
predictions = predict(X, beta)

print("Learned coefficients:")
print(f"Beta_0 (Intercept)      : {beta[0]}")
print(f"Beta_1 (Square_Footage) : {beta[1]}")
print(f"Beta_2 (Num_Bedrooms)   : {beta[2]}")
print(f"Beta_3 (Num_Bathrooms)  : {beta[3]}")
```

```python
for i in range(10):
  print(f"Sample:{i}----> Predicted={predictions[i]}----> Actual={y[i]}")
```
