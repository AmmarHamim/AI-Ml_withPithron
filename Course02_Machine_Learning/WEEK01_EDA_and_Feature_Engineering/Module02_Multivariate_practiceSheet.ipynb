**1. Countplot এবং Barplot এর মধ্যে পার্থক্য কী? কখন কোনটা use করবে?**

Seaborn countplot ব্যবহার হয় কোন ক্যাটেগরি কতবার আছে তা দেখানোর জন্য। শুধু categorical

Seaborn barplot ব্যবহার হয় কোন ক্যাটেগরির সাথে numerical value-এর average (বা aggregate) দেখাতে। categorical + numerical

**2. Scatterplot দেখে কীভাবে বুঝবে দুটো variable এর মধ্যে সম্পর্ক আছে কিনা?**

যদি পয়েন্টগুলো একটা নির্দিষ্ট দিক ধরে সাজানো থাকে, তাহলে সম্পর্ক আছে বলা যায়। যেমন:

📈 Positive relationship (ধনাত্মক সম্পর্ক): এক variable বাড়লে অন্যটাও বাড়ে → পয়েন্টগুলো নিচ থেকে উপরে ডানদিকে উঠে যায়

📉 Negative relationship (ঋণাত্মক সম্পর্ক): একটা বাড়লে অন্যটা কমে → পয়েন্টগুলো উপরের বাম থেকে নিচের ডানদিকে নামে

❌ No relationship (কোনো সম্পর্ক নেই): পয়েন্টগুলো এলোমেলোভাবে ছড়ানো থাকে, কোনো clear pattern থাকে না

**3. Y Data Profiling কেন manually EDA করার চেয়ে সুবিধাজনক? কী কী information এক জায়গায় পাওয়া যায়?**

Y Data Profiling manually EDA করার চেয়ে সুবিধাজনক কারণ এটি খুব দ্রুত পুরো dataset-এর একটি complete summary report তৈরি করে দেয়, যেখানে হাতে-কলমে আলাদা আলাদা কোড লিখে সব কিছু বের করতে হয়।

✅ কেন Y Data Profiling বেশি সুবিধাজনক?
⚡ Time-saving: এক লাইনের কোডেই পুরো EDA report তৈরি হয়
📊 Automated analysis: manually mean, missing value, distribution ইত্যাদি আলাদা করে লিখতে হয় না
🔍 Quick data understanding: dataset সম্পর্কে overall ধারণা খুব দ্রুত পাওয়া যায়
⚠️ Error reduction: manual calculation বা oversight কম হয়
📌 এক জায়গায় কী কী information পাওয়া যায়?

Y Data Profiling report সাধারণত নিচের তথ্যগুলো একসাথে দেয়:

🧾 Dataset overview (rows, columns, memory usage)
❌ Missing values analysis (কোন column-এ কত missing data আছে)
🔢 Data types (numeric, categorical, etc.)
📊 Descriptive statistics (mean, median, std, min, max)
📈 Distributions of variables (histogram, frequency)
🔗 Correlations between variables
⚠️ Duplicate rows detection
📉 Outliers indication
🧠 Unique values per column
🔍 Data quality warnings

**4. নিচের code টি একটি Countplot তৈরি করবে যেখানে passenger class অনুযায়ী count দেখাবে এবং Gender দিয়ে আলাদা করবে।**

*Hint: x = ticket class column, hue = gender column*

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("titanic")
df

sns.countplot(data=df, x='pclass', hue='sex')

plt.title("Passenger Class by Gender")
plt.show()
```

**5. নিচের code টি City অনুযায়ী average Salary দেখাবে।**

*Hint: Cat-Num relationship দেখানোর plot, x = category, y = number*

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame({
    "City": ["Dhaka", "Dhaka", "CTG", "CTG", "Sylhet"],
    "Salary": [50000, 60000, 45000, 55000, 40000]
})

sns.boxplot(data=df, x='City', y='Salary')

plt.title("Average Salary by City")
plt.show()
```

**6. নিচের code টি Age column এর smooth distribution curve দেখাবে।**

*Hint: Histogram এর smooth version যে plot দেখায়*

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("titanic")

sns.kdeplot(data=df, x='age')

plt.title("Age Distribution")
plt.show()
```

**7. নিচের code টি Height এবং Weight এর মধ্যে সম্পর্ক দেখাবে।**

*Hint: দুটো numerical variable এর মধ্যে relationship দেখানোর plot*

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame({
    "Height": [150, 160, 170, 175, 180],
    "Weight": [50, 60, 70, 75, 80]
})

sns.scatterplot(data=df, x='Height', y='Weight')

plt.title("Height vs Weight")
plt.show()
```
