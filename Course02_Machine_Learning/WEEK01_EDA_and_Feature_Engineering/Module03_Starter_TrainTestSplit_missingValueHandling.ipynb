### 3.6 Numerical Missing Value Imputation with Pandas

```python
X_train.isnull().sum()
```

```python
X_test.isnull().sum()
```

```python
#mean impute
age_mean=X_train['Age'].mean()

X_train['age_mean_imputed']=X_train['Age'].fillna(age_mean)
X_test['age_mean_imputed']=X_test['Age'].fillna(age_mean)

X_test
```

```python
age_mean
```

```python
#median impute

age_median = X_train['Age'].median()


X_train['age_median_imputor'] = X_train['Age'].fillna(age_median)

X_test['age_median_imputor'] = X_test['Age'].fillna(age_median)

X_train
```

```python
X_train.isnull().sum()
```

```python
sns.kdeplot(data=X_train,x='age_mean_imputed')
```

```python
sns.kdeplot(data=X_train,x='age_median_imputor')
```

### 3.7 Numerical Missing Values Imputation using Simple Imputer

```python
from sklearn.impute import SimpleImputer

age_imputer=SimpleImputer(missing_values=np.nan,strategy='mean')

age_imputer.fit(X_train[['Age']])
X_train['Age']=age_imputer.transform(X_train[['Age']]).ravel()
X_test['Age']=age_imputer.transform(X_test[['Age']]).ravel()
```

```python
X_train.isnull().sum()
```

```python
X_train.drop(['age_mean_imputed','age_median_imputor'],axis=1,inplace=True)
```

```python
X_test.drop(['age_mean_imputed','age_median_imputor'],axis=1,inplace=True)
```

### 3.8 Categorical Value Imputation with Mode Value

```python
embarked_imputer=SimpleImputer(strategy='most_frequent') # by default missing_values=nan

embarked_imputer.fit(X_train[['Embarked']])

X_train['Embarked']=embarked_imputer.transform(X_train[['Embarked']]).ravel()
X_test['Embarked']=embarked_imputer.transform(X_test[['Embarked']]).ravel()
```

```python
X_train.isnull().sum()
```

```python
X_test.isnull().sum()
```

### 3.9 Categorical Value Imputation with Missing string and Indicator

```python
sns.countplot(data = X_train , x=X_train['Cabin'])
```

```python
cabin_imputer=SimpleImputer(strategy='constant',fill_value='Missing',add_indicator=True)

cabin_imputer.fit(X_train[['Cabin']])

X_train[['Cabin','Cabin_missing_indicator']]=cabin_imputer.transform(X_train[['Cabin']])
X_test[['Cabin','cabin_missing_indicator']] = cabin_imputer.transform(X_test[['Cabin']])
```
