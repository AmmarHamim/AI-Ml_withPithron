## 03 label encoding

```python
df = sns.load_dataset('tips')

df
```

```python
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder

le=LabelEncoder()

df['sex']=le.fit_transform(df['sex'])

df
```

```python
df['smoker']=le.fit_transform(df['smoker'])

df
```

```python
def tip_cat(tip):
  if tip<3:
    return 'Low'
  elif tip<6:
    return 'Medium'
  else:
    return 'High'

df['Tip_Category']=df['tip'].apply(tip_cat)
```

```python
df
```

```python
tip_enc=OrdinalEncoder(categories=[['Low','Medium','High']])

df['Tip_Category_encoded']=tip_enc.fit_transform(df[['Tip_Category']])

df
```
