```python
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder

fuel_enc=OrdinalEncoder(categories=[['gas','diesel']],handle_unknown='use_encoded_value',unknown_value=-1)
X_train['fueltype']=fuel_enc.fit_transform(X_train[['fueltype']]).ravel()
X_test['fueltype']=fuel_enc.transform(X_test[['fueltype']]).ravel()
```
