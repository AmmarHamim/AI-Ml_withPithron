```python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df.head()
```

```python
df.shape
```

```python
df.drop(columns=['id', 'Unnamed: 32'], inplace= True)
```

```python
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 1:], df.iloc[:, 0], test_size=0.2,random_state=42)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

```python
X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test)
y_train_tensor = torch.from_numpy(y_train).to(torch.float64)
y_test_tensor = torch.from_numpy(y_test).to(torch.float64)
```

```python
X_train_tensor.shape,y_train_tensor.shape
```

```python
X_test_tensor.dtype,y_test_tensor.dtype
```

# Defining the model

```python
import torch.nn as nn
```

```python
# class MySimpleNN()
  #initialize

  #forward function

  #loss function


class MySimpleNN(nn.Module):

  def __init__(self, num_features):
    super().__init__()

    self.layer=nn.Linear(num_features,1,dtype=torch.float64)
    self.activation=nn.Sigmoid()
    self.criterion=nn.BCELoss()

  def forward(self,features):
    z=self.layer(features)
    out=self.activation(z)
    return out

  def loss_function(self,y_pred,y):
    loss=self.criterion(y_pred,y)
    return loss
```

```python
learning_rate = 0.1
epochs = 25
```

# Training Pipeline

```python
# create model
torch.manual_seed(42)
model=MySimpleNN(X_train_tensor.shape[1])

#create optimizer
optim=torch.optim.SGD(model.parameters(),lr=learning_rate)

#epoch loop
for epoch in range(epochs):
  # forward pass
  y_pred=model.forward(X_train_tensor)

  # loss
  loss=model.loss_function(y_pred,y_train_tensor.reshape(-1,1))

  # backward pass
  loss.backward()

  # param update
  optim.step()

  # zero grad

  optim.zero_grad()

  # print loss in each epoch
  print(f'Epoch: {epoch + 1}, Loss: {loss.item()}')
```

```python
model.layer.weight,model.layer.bias
```

# Evaluation

```python
# model evaluation
with torch.no_grad():
  y_pred = model.forward(X_test_tensor)
  y_pred = (y_pred > 0.6).float()
  accuracy = (y_pred == y_test_tensor).float().mean()
  print(f'Accuracy: {accuracy.item()}')
```
