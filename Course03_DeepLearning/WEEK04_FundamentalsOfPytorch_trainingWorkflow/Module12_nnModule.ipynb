```python
import torch
import torch.nn as nn
```

#version1

```python
class Model(nn.Module):
  def __init__(self,num_features):
    super().__init__()

    self.linear = nn.Linear(num_features,1)
    self.sigmoid = nn.Sigmoid()


  def forward(self,features):
    out = self.linear(features)
    out = self.sigmoid(out)

    return out
```

```python
# Create dataset

features = torch.rand(10,5)
#create model

model = Model(features.shape[1])
model(features)
```

```python
model.linear.weight
```

# Version2

```python
class Model(nn.Module):
  def __init__(self,num_features):
    super().__init__()

    #layer 1

    self.linear1 = nn.Linear(num_features,3)
    self.relu = nn.ReLU()

    #layer 2
    self.linear2 = nn.Linear(3,1)
    self.sigmoid = nn.Sigmoid()


  def forward(self,features):
    z1 = self.linear1(features)
    out = self.relu(z1)

    z2 = self.linear2(out)
    out = self.sigmoid(z2)

    return out
```

```python
# Create dataset

features = torch.rand(10,5)
#create model

model = Model(features.shape[1])
model(features)
```

# version3

```python
class Model(nn.Module):
  def __init__(self,num_features):
    super().__init__()

    self.network=nn.Sequential(
        nn.Linear(num_features,3), #index0
        nn.ReLU(),
        nn.Linear(3,1),
        nn.Sigmoid()
        )

  def forward(self,features):

    out=self.network(features)

    return out
```

```python
# Create dataset

features = torch.rand(10,5)
#create model

model = Model(features.shape[1])
model(features)
```

```python
model.network[0].weight
```

```python
print("Layer 1 weights:")
print(model.network[0].weight)

print("\nLayer 1 bias:")
print(model.network[0].bias)

print("\nLayer 2 weights:")
print(model.network[2].weight)

print("\nLayer 2 bias:")
print(model.network[2].bias)
```
