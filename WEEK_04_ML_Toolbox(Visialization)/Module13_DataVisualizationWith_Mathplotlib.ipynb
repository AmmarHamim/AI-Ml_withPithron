# Bar Chart
## for categorial data

```python
data = {
    'Student': [ 'B', 'C', 'D', 'E', 'F', 'G', 'H'],
    'Gender': [ 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male'],
    'StudyHours': [ 2, 5, 3, 7, 4, 8, 1],
    'Attendance': [ 60, 88, 70, 95, 80, 98, 55],
    'Grade': [ 'Fail', 'Pass', 'Fail', 'Pass', 'Pass', 'Pass', 'Fail']
}

df = pd.DataFrame(data)
```

```python
df
```

```python
# gender wise

gender_grp=df.groupby("Gender").size()

gender_grp
```

```python
print(gender_grp.index)
gender_grp.values
```

```python
plt.bar(gender_grp.index,gender_grp.values)
plt.xlabel('Gender')
plt.ylabel('Count')
plt.title('Bar Chart of Gender among students')
```

```python
#  Bar chart on real data

df = pd.read_csv('student_data (1).csv')

df
```

```python
status=df.groupby('CompletionStatus').size()

status
```

```python
plt.bar(status.index,status.values,edgecolor='black',color=['green','orange','red'])
plt.xlabel('Completion Status')
plt.ylabel('Count')
plt.title('Bar Chart of Completion Status')
plt.show()
```

```python
status = df.groupby('Location').size()

plt.bar(status.index,status.values,edgecolor='black')

plt.xlabel('Location')
plt.ylabel('Count')
plt.title('Bar Chart of Location of Students')
plt.show()
```

# Pie Chart

```python
status=df.groupby('CompletionStatus').size()

plt.pie(status,labels=status.index,autopct='%1.1f%%')
plt.title('Pie Chart of Completion Status')
plt.show()
```

```python
status=df.groupby('CompletionStatus').size()

plt.pie(status,labels=status.index,autopct='%1.1f%%',explode=(0.1,0,0),shadow=True)
plt.title('Pie Chart of Completion Status')
plt.show()
```

3. Handle Missing Values
Columns: Age, Salary
 Tasks:
Count missing values


Fill missing Age with mean age


Drop rows where Salary is missing

```python
df[['Age','Salary']].isnull().sum()
```

```python
df['Age']=df['Age'].fillna(df['Age'].mean())

df
```

```python
df.dropna(subset='Salary')
```

4. Apply Statistical Functions
Columns: Marks1, Marks2, Marks3
 Tasks:

Calculate mean, max for each subject

Compute total marks per student


Find standard deviation of total marks


```python
mean_values = df[["Marks1", "Marks2", "Marks3"]].mean()

mean_values
```

```python
max_values = df[["Marks1", "Marks2", "Marks3"]].max()

max_values
```

```python
df['Total_Marks']=df[['Marks1','Marks2','Marks3']].sum(axis=1)

df
```

```python
df['Total_Marks'].std()
```
